# 데이터 EDA

```python
# 가장 많이 구매 된 제품명 & 대분류
# 거래시간대 별 구매 상품
# 점포별 팔린 제품의 차이가 있을 것이다. (T검정, 카이검정)
# 점포와 연월별 판매량 추이
# 장바구니 분석
```

```python
import pandas as pd
df1=pd.read_csv('/content/구매내역정보.csv',encoding='cp949')
df1
```

```python
gb=df1.groupby(['점포ID','상품중분류명'])['구매수량'].sum()
# 각 점포별로 구매수량이 가장 높은 상품중분류명의 인덱스(점포ID, 상품중분류명)를 찾습니다.
best_seller_indices = gb.groupby(level='점포ID').idxmax() # 최대값의 인덱스 반환

# 해당 인덱스를 사용하여 gb Series에서 가장 많이 팔린 제품의 정보(제품명, 구매수량)를 가져옵니다.
best_sellers_info = gb.loc[best_seller_indices]

display(best_sellers_info) #더 보기 좋게 출력하기 위해 사용
```
